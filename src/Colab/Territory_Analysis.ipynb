```python
w_kernel = weights.distance.Kernel.from_dataframe(tuatha_nw_1600, fixed=False, k=4, function='triangular', radius=None)
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create the plot with larger figure size
plt.figure(figsize=(20, 20))

# Convert the spatial weights matrix to a networkx graph
G = w_kernel.to_networkx()

# Create a layout for the nodes
pos = {index: (row['CENTROID_X'], row['CENTROID_Y']) for index, row in tuatha_nw_1600.iterrows()}

# Draw the nodes with varying sizes based on degree centrality and color by province
node_sizes = [G.degree(node) * 200 for node in G.nodes()]  # Increased from 200 to 500
node_colors = ['red' if row['PROVINCE']=='Ulster' else 'blue' for _, row in tuatha_nw_1600.iterrows()]

# Draw nodes
nx.draw_networkx_nodes(G, pos, 
                      node_size=node_sizes,
                      node_color=node_colors,
                      alpha=0.7)

# Draw edges with reduced alpha for clarity
nx.draw_networkx_edges(G, pos, 
                      alpha=0.2,
                      edge_color='gray')

# Add labels with territory names
labels = {index: row['ENGLISH'] for index, row in tuatha_nw_1600.iterrows()}
nx.draw_networkx_labels(G, pos, labels, font_size=8)

# Add a title and legend
plt.title('Territory Connections Network\nNode size represents number of connections\nRed=Ulster, Blue=Connacht', 
          pad=20, fontsize=14)

plt.axis('off')
plt.show()

# Print summary statistics
print("\nNetwork Analysis Results:")
print("-----------------------")
print(f"Total territories: {len(G.nodes())}")
print(f"Total connections: {len(G.edges())}")
print("\nMost connected territories:")
degrees = dict(G.degree())
for territory, degree in sorted(degrees.items(), key=lambda x: x[1], reverse=True)[:5]:
    print(f"{tuatha_nw_1600.iloc[territory]['ENGLISH']}: {degree} connections")
```

```python
# Degree Centrality
degree_centrality = nx.degree_centrality(G)

# Closeness Centrality
closeness_centrality = nx.closeness_centrality(G)

# Betweenness Centrality
betweenness_centrality = nx.betweenness_centrality(G)

# Convert the metrics to dataframes for easier handling
degree_df = pd.DataFrame.from_dict(degree_centrality, orient='index', columns=['Degree Centrality'])
closeness_df = pd.DataFrame.from_dict(closeness_centrality, orient='index', columns=['Closeness Centrality'])
betweenness_df = pd.DataFrame.from_dict(betweenness_centrality, orient='index', columns=['Betweenness Centrality'])

# Join the metrics to the original GeoDataFrame
tuatha_nw_1600 = tuatha_nw_1600.join(degree_df)
tuatha_nw_1600 = tuatha_nw_1600.join(closeness_df)
tuatha_nw_1600 = tuatha_nw_1600.join(betweenness_df)

# Print the territories with the highest degree centrality
print(tuatha_nw_1600.nlargest(5, 'Degree Centrality'))

# Print the territories with the highest closeness centrality
print(tuatha_nw_1600.nlargest(5, 'Closeness Centrality'))

# Print the territories with the highest betweenness centrality
print(tuatha_nw_1600.nlargest(5, 'Betweenness Centrality'))
```
